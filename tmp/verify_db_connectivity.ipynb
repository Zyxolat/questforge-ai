```python
# Verify Local Docker/Postgres Connectivity
This notebook checks the local Docker PostgreSQL container and environment configuration for the backend.
</VSCode.Cell>
<VSCode.Cell id="2" language="python">
import subprocess
from pathlib import Path

workspace = Path('/home/zyxolat/Desktop/QuestForge AI')

commands = [
    ['docker', 'ps', '--filter', 'name=questforge-postgres', '--format', '{{.ID}} {{.Names}} {{.Status}} {{.Ports}}'],
    ['docker', 'inspect', '-f', '{{range .Config.Env}}{{println .}}{{end}}', 'questforge-postgres'],
]

for cmd in commands:
    print('Running:', ' '.join(cmd))
    try:
        output = subprocess.run(cmd, cwd=workspace, capture_output=True, text=True, check=True)
        print(output.stdout)
    except subprocess.CalledProcessError as exc:
        print('ERROR:', exc.returncode)
        print(exc.stdout)
        print(exc.stderr)
</VSCode.Cell>
```
